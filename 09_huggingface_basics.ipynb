## CELL1

```python
from transformers import AutoTokenizer
from transformers import AutoModel
```

## CELL2-> LOAD GPT2

```python
tokenizer = AutoTokenizer.from_pretrained(
    "gpt2"
)

model = AutoModel.from_pretrained(
    "gpt2"
)

# What Gets Downloaded?

# Tokenizer:

# Vocabulary
# BPE Rules

# Model:

# Weights
# Configuration
# Architecture

# This is basically:

# Your GPT From Scratch

# except trained by OpenAI.
```

## CELL3-> PREPARE INPUT

```python
text = "I love transformers"

inputs = tokenizer(
    text,
    return_tensors="pt"
)

print(inputs)

# What Is return_tensors="pt" ?

# Without:

# tokenizer.encode()

# returns:

# [40,1842,6121]

# Python list.

# With:

# return_tensors="pt"

# returns:

# tensor([[40,1842,6121]])

# PyTorch tensor.
```

## Cell 4

Pass through model.

```python
outputs = model(
    **inputs
)
print(outputs)
print(
    outputs.last_hidden_state.shape
)
# This is the final Transformer output.

# So with:

# AutoModel

# you stop at:

# Input Text
 
# Tokenizer
 
# Input IDs
 
# Embeddings
 
# Transformer
 
# last_hidden_state

```

## Cell 5

Inspect one token.

```python
print(
    outputs.last_hidden_state[0,0]
)

# This is the learned representation for:

# "I"

# after passing through the transformer.


# Important Connection To Previous Phases

# Earlier:

# Phase 3

# Token Embedding
# ↓
# Attention
# ↓
# FFN
# ↓
# Hidden State

# Now:

# outputs.last_hidden_state

# IS that hidden state.

# You're looking at the actual output of a pretrained GPT-2 transformer.
```

#### Remember:

This is where everything from GPT implementation comes together.

## Cell 1

```python
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM
)
```

## Cell 2

Load GPT-2

```python
tokenizer = AutoTokenizer.from_pretrained(
    "gpt2"
)

model = AutoModelForCausalLM.from_pretrained(
    "gpt2"
)

# What Changed?

# Earlier:

# AutoModel

# Now:

# AutoModelForCausalLM

# Meaning:

# Hidden States
# ↓
# LM Head
# ↓
# Vocabulary Logits

# are included.
```

## Cell 3

Prepare Input

```python
text = "I love"

inputs = tokenizer(
    text,
    return_tensors="pt"
)
inputs["input_ids"].shape
```

## Cell 4

Forward Pass

```python
outputs = model(
    **inputs
)
print(
    outputs.logits.shape
)
print(outputs)
```

## Cell 5

Get Last Token Prediction

```python
next_token_logits = outputs.logits[
    :,
    -1,
    :
]
print(next_token_logits)

# Meaning:

# Scores for next token

# after:

# I love
```

## Cell 6

Convert To Probabilities

```python
import torch

probs = torch.softmax(
    next_token_logits,
    dim=-1
)
probs.shape
print(probs)
```

## Cell 7

Choose Most Likely Token

```python
next_token = torch.argmax(
    probs,
    dim=-1
)
print(next_token) # gives idnex of the largest probablity
```

### Decode

```python
print(
    tokenizer.decode(
        next_token
    )
)

# What Just Happened?

# We manually performed:

# Prompt
# ↓
# Tokenizer
# ↓
# Transformer
# ↓
# LM Head
# ↓
# Logits
# ↓
# Softmax
# ↓
# Argmax
# ↓
# Next Token

# Exactly what GPT does.

# But We Don't Usually Do This

# Hugging Face provides:

# generate()
```

## Cell 8

Text Generation

```python
output = model.generate(
    **inputs,
    max_new_tokens=20
)
print(output.shape)
```

### Decode

```python
generated_text = tokenizer.decode(
    output[0]
)

print(generated_text)

# What Happens Internally?
# I love
# ↓
# Predict Token
# ↓
# Append
# ↓
# Predict Next Token
# ↓
# Append
# ↓
# Repeat

# Exactly like our:

# generate()

# method from GPT-from-scratch.
```

## First Embedding Model

```python
from sentence_transformers import SentenceTransformer

# 1. Load a pretrained Sentence Transformer model
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

# What Happens?

# Hugging Face downloads:

# Tokenizer

# Transformer

# Weights

# Embedding Head

# This model is specifically trained for:

# Semantic Search
```

## Cell 3

Create Embedding

```python
sentence = (
    "I love machine learning"
)

embedding = model.encode(
    sentence
)
print(
    embedding.shape
)

s1 = "I love AI"

s2 = "I enjoy machine learning"

s3 = "Pizza tastes great"
emb1 = model.encode(s1)

emb2 = model.encode(s2)

emb3 = model.encode(s3)
```

## CELL4 COSINE SIMILARITY

```python
from sklearn.metrics.pairwise import (
    cosine_similarity
)
sim12 = cosine_similarity(
    [emb1],
    [emb2]
)

sim13 = cosine_similarity(
    [emb1],
    [emb3]
)

print(sim12)

print(sim13)
```
